```python
from data_dallion_framework.Extractors.APIExtractor import APIAutomation
from data_dallion_framework.Common import (
    JsonDataMapper,
    FileNameGenerator,
    OrchestrationProcess,
)
```

```python
temp_dict = dict()
temp_dict["method"] = "GET"
temp_dict["type"] = "RESPONSE"
temp_dict["ssl_verify"] = False
temp_dict["url"] = "https://priceapi.moneycontrol.com/techCharts/mcApp/history"
temp_dict["params"] = {"symbol": "$symbol$", "resolution": "1D", "from": "1577836800", "to": "1764720000"}
temp_dict['body_values'] = {'$symbol$':['GANESHHOU','MMTC']}
temp_dict['key_to_add_to_data'] = "$symbol$"
```

```python
api_automation = APIAutomation(config=[temp_dict]).execute_workflow()
```

```python
api_responses = api_automation['values_based_response']
```

```python
column_mapping = {
    "symbol":"$.static_value",
    "open_price":"data[*].o",
    "close_price":"data[*].c",
    "high_price":"data[*].h",
    "low_price":"data[*].l",
    "timestamp":"data[*].t",
}
```

```python
results = []
for api_response in api_responses:
    results.extend(JsonDataMapper.JsonDataMapper(column_mapping,api_response).get_mapped_data())
```
